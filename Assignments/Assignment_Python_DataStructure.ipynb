```python
#Problem 1

p = 500
q = 3
d = .15
t = .18
_curr = "\u20B9"

#1. Calculate total amount before discount. 
total_amt = p*q
print(f"Total amount is {_curr}{total_amt}")

#2. Calculate discount amount. 
dis_amt = (total_amt * d)
print (f"Discount amount is {_curr}{round(dis_amt,2)}")

#3. Calculate amount after discount. 
amt_dis = round((total_amt - dis_amt),2) 
print (f'Amount after discount is {_curr}{amt_dis}')

#4. Add tax to calculate final bill. 
tax_amt = round(amt_dis * t,2)
#print(amt_tax)
amt_tax = round(amt_dis + tax_amt,2)
print (f'Amount with tax is {_curr}{amt_tax}')

#5. Print final amount. 
print(f"Final amount with tax is {_curr}{amt_tax}")
```

```python
#Problem 6 – Reverse String 
customer_id = "CUST10234"
#Slicing
print(f"Customer Id in reverse order using Slicing is {customer_id[::-1]}")

#using loop
str_reverse=""
for c in customer_id:
    str_reverse = c + str_reverse
print(f"Customer Id in reverse order using loop is {str_reverse}")
```
